# 01 — Check Burn Label Outputs (Phase 1)

Quick visual / numerical sanity check of the GeoTIFFs produced by `scripts/task1/phase_01_burn_labels/extract_burn_labels.py`.

Expected inputs in `data/raw/satellite_labels/`:
- `<fire_id>_BurnedArea.tif` (MODIS, 500 m)
- `<fire_id>_Severity.tif` (Landsat dNBR, 30 m)


```python
from pathlib import Path

import numpy as np
import rasterio
from rasterio.plot import show
import matplotlib.pyplot as plt

LABELS_DIR = Path('../data/raw/satellite_labels')
sorted(LABELS_DIR.glob('*.tif'))
```

```python
# Pick one fire ID and inspect its two rasters.
FIRE_ID = 'REPLACE_WITH_FIRE_ID'

burn_path = LABELS_DIR / f'{FIRE_ID}_BurnedArea.tif'
sev_path  = LABELS_DIR / f'{FIRE_ID}_Severity.tif'
print('Burned area:', burn_path, burn_path.exists())
print('Severity   :', sev_path,  sev_path.exists())
```

```python
# Open and inspect basic raster metadata.
with rasterio.open(burn_path) as src:
    print('Burned-area metadata:')
    print('  CRS    :', src.crs)
    print('  Shape  :', (src.height, src.width))
    print('  Bounds :', src.bounds)
    burn = src.read(1)

with rasterio.open(sev_path) as src:
    print('Severity metadata:')
    print('  CRS    :', src.crs)
    print('  Shape  :', (src.height, src.width))
    print('  Bounds :', src.bounds)
    sev = src.read(1)

print('\nBurned pixel count :', int(np.sum(burn > 0)))
print('Severity histogram :', np.unique(sev, return_counts=True))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(burn, cmap='Reds')
axes[0].set_title(f'{FIRE_ID} — Burned area (MODIS)')
axes[1].imshow(sev, cmap='YlOrRd', vmin=0, vmax=4)
axes[1].set_title(f'{FIRE_ID} — Severity (Landsat dNBR)')
for ax in axes:
    ax.axis('off')
plt.tight_layout()
plt.show()
```
